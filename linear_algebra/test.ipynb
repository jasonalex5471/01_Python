```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 定义一个矩阵 A (代表一种线性变换：把空间拉伸)
# 试试修改这里的数字，看看图像的变化
A = np.array([[2, 1], 
              [0, 1]])

# 2. 定义原始基向量 i 和 j
i = np.array([1, 0])
j = np.array([0, 1])

# 3. 计算变换后的基向量
new_i = A @ i
new_j = A @ j

# 4. 绘图
plt.figure(figsize=(5, 5))
plt.quiver([0, 0], [0, 0], [i[0], j[0]], [i[1], j[1]], color='blue', angles='xy', scale_units='xy', scale=1, label='Original')
plt.quiver([0, 0], [0, 0], [new_i[0], new_j[0]], [new_i[1], new_j[1]], color='red', angles='xy', scale_units='xy', scale=1, label='Transformed')

plt.xlim(-1, 4)
plt.ylim(-1, 4)
plt.grid(True)
plt.legend()
plt.title("Linear Transformation of Basis Vectors")
plt.show()
```
